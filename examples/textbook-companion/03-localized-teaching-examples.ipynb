# Localized teaching examples

This notebook regenerates localized teaching examples and checks locale/accessibility coverage.

```python
import json
import pathlib
import subprocess

command = "go run ./cmd/patchline localized-teaching-examples --spec examples/localized-teaching-examples.json --root . --out results/generated/textbook-companion/localized-teaching-examples --json"
subprocess.run(command, shell=True, check=True)
report = json.loads(pathlib.Path("results/generated/textbook-companion/localized-teaching-examples/localized-teaching-examples.json").read_text())
assert report["ok"] is True
assert report["summary"]["locales_covered"] >= 2
```
